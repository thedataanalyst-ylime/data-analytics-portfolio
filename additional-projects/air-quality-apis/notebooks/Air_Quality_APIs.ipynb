Now for **Step 2: Doing this many times for different cities and joining them**

Let's create this dataframe for several cities. Write a loop through the below cities that creates a list of dictionaries in the same format as the one above. In the next step, you'll call merge this list to join all the rows into one large dataframe.

Your loop will involve two API calls: one for weather, one for air pollution.

```python
# Create a list of dictionaries for the 10 cities

import requests

cities = [
    'Chicago,IL,US',
    'Boston,MA,US',
    'Austin,TX,US',
    'New York City,NY,US',
    'Atlanta,GA,US',
    'Los Angeles,CA,US',
    'Seattle,WA,US',
    'Minneapolis,MN,US',
    'Denver,CO,US',
    'Washington,DC,US'
]

city_data_list = []

for city in cities:
    
    #1. Weather API call
    weather_url = f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid={OPENWEATHER_API_KEY}&units={units}"
    weather_response = requests.get(weather_url)
    weather_response.raise_for_status()
    weather_data = weather_response.json()
    
    #Get coordinates from weather response
    latitude = weather_data['coord']['lat']
    longitude = weather_data['coord']['lon']
    
    #2. Air Pollution API call
    air_url = f"http://api.openweathermap.org/data/2.5/air_pollution?lat={latitude}&lon={longitude}&appid={OPENWEATHER_API_KEY}&units=metric"
    air_response = requests.get(air_url)
    air_response.raise_for_status()
    air_data = air_response.json()
    
    #3. Create dictionary in same format as Chicago
    city_dict = {
        'city': weather_data['name'],
        
        'coordinates': {
            'lat': latitude,
            'lon': longitude
        },
        
        'weather_main': weather_data['main'],
        
        'air_quality': {
            'aqi': air_data['list'][0]['main']['aqi'],
            'components': air_data['list'][0]['components']
        }
    }
    
    # Add dictionary to the list
    city_data_list.append(city_dict)

# View the list
city_data_list
```

It's time to join all of these dictionaries to create one large dataframe of 10 rows of weather data! To do this, simply call `pd.DataFrame()` on your list of dictionaries and you should be good to go!

```python
# Convert list into 1 dataframe

import pandas as pd

cities_df = pd.json_normalize(city_data_list)

# Clean column names
cities_df.columns = cities_df.columns.str.replace('.', '_', regex=False)

cities_df
```

**What is PM2.5?**
- PM2.5 refers to fine particulate matter that is 2.5 micrometers or smaller in diameter. <br>
It is about 30 times smaller than the width of a human hair. <br>
And because it is so tiny, it can travel deep into the lungs, enter the bloodstream and affect multiple organs in the body.<br>

- PM2.5 is commonly produced by vehicle emissions, industrial activity, burning of fossil fuels, construction dust, wildfires, household fuel burning. <br>

**Why PM2.5?** <br>
- PM2.5 is selected for this analysis because it is one of the most harmful and widely monitored air pollutants. <br>
Due to its fine particle size, it can penetrate deep into the lungs and negatively impact human health. <br> 

- PM2.5 also reflects urban activity patterns clearly, making it suitable for analyzing how daily commuting behavior affects air quality.<br>

**Why is PM2.5 important?** <br>
- PM2.5 is considered one of the most harmful air pollutants because prolonged exposure is linked to:
    * respiratory diseases
    * asthma
    * cardiovascular diseases
    * lung cancer
    * premature death


**STEP 1: Import required libraries and account key. Identity endpoint & sensor_id.**

```python
import requests
import pandas as pd
import plotly.express as px
import time

#Refresh Jupyter cache. Prevents use of old or previous imports / caches.
import importlib
import config

importlib.reload(config)

# Account Key is stored in a config file
from config import AIRQUALITY_API_KEY

url = 'https://api.openaq.org/v3'
sensor_id = 23534
```

**STEP 2: Call the API and do Error Handling.**

```python
# 1. AUTHENTICATION HEADER
# Header that will be passed to the OpenAQ API

headers = {
    'X-API-Key': AIRQUALITY_API_KEY
}

# 2. API ENDPOINT & QUERY PARAMETERS TO GET THE MEASUREMENTS AGGREGATED FROM HOUR TO DAY OF WEEK BY SENSOR ID
# From the Sensors Resources:
# API Endpoint using the sensor_id

# From the Measurements Resources:
# hourofday: The hourofday aggregator returns the average (mean) value for each hour in a 24 hour day, across the selected time range. 
# This allows analyzing air quality trends on an hourly basis, offering insights into daily patterns and peak pollution times, such as diurnal patterns.

url = f"{url}/sensors/{sensor_id}/hours/hourofday"

# These parameters define the scope of the API request:
# a. datetime_from / datetime_to: filters data to April 2026
# b. limit: maximum number of records returned (up to 1000)

params = {
    'datetime_from': '2026-04-01',
    'datetime_to': '2026-04-30',
    'limit': 1000
}

# 3. API REQUEST + ERROR HANDLING
# This calls the API via requests and safely handles possible errors.

try:
    # Send GET request to OpenAQ API with headers and parameters
    response = requests.get(url, headers=headers, params=params)

    # Pause briefly to avoid hitting API rate limits
    time.sleep(1)

    # Raise an error if the response status is not successful
    response.raise_for_status()

    # Parse JSON response and extract the 'results' data
    data = response.json()['results']

# 4. HANDLE HTTP ERRORS
# Catches API-related errors (e.g., 401, 404, 408, 429)

except requests.exceptions.HTTPError as e:
    print("HTTP error:", response.status_code, response.text)
    data = [] # reset to empty dataset

# 5. HANDLE OTHER ERRORS
# Catches unexpected errors (e.g., connection issues, parsing errors)

except Exception as e:
    print("Other error:", e)
    data = [] # reset to empty dataset
```

```python
data
```

**STEP 3: Extract the data, save in a Dataframe and analyze data.**

```python
# 1. INITIALIZE EMPTY LIST
# Create an empty list to store cleaned and structured records.
# Each record will represent one hour's average PM2.5 value.

hour_data = []

# 2. TRANSFORM RAW API DATA
# Loop through each item in the API response ('data')
# and extract only the relevant fields needed for analysis.

for item in data:
    hour_data.append({
         # Extract the hour (0–23) from the 'period' label and convert to integer
        "hour": int(item["period"]["label"]),

        # Extract the average PM2.5 value for that hour
        "pm25_value": item["value"],

        # Extract the unit of measurement (µg/m³: micrograms per cubic meter)
        "unit": item["parameter"]["units"]
    })

# 3. CREATE DATAFRAME + SORT
# Convert the list of dictionaries into a pandas DataFrame
# and sort it by hour to ensure proper chronological order for plotting.

air_quality_df = pd.DataFrame(hour_data).sort_values("hour")
air_quality_df
```

```python
hour_data
```

**STEP 4: Visualize the data.**

```python
# 1. IDENTIFY KEY DATA POINTS (PEAK & LOWEST)
# Find the hour with the highest PM2.5 value (peak pollution)
peak = air_quality_df.loc[air_quality_df["pm25_value"].idxmax()]

# Find the hour with the lowest PM2.5 value (cleanest air)
low = air_quality_df.loc[air_quality_df["pm25_value"].idxmin()]

# 2. DESIGN THE DATA VISUALIZATION
# Plot a LINE CHART showing PM2.5 levels across hours of the day
# Markers are added to highlight each data point
fig = px.line(
    air_quality_df,
    x="hour",
    y="pm25_value",
    markers=True,
    title="Rush Hour Pollution: When Delhi’s Air Is Most Dangerous<br>"
          "<sup>PM2.5 levels peak during commuting hours, showing the impact of daily urban activity (April 2026)</sup>"
)

# 3. ADD ANNOTATION FOR POLLUTION PEAK HOURS
# Highlight the hour with the highest PM2.5 level
fig.add_annotation(
    x=peak["hour"],
    y=peak["pm25_value"],
    text=f"<b>Peak: {peak['hour']}h</b>",
    showarrow=False,
    yshift=-20,     # move UP
    xshift=10      # move RIGHT
)

# 4. ADD ANNOTATION FOR LOWEST POLLUTION
# Highlight the hour with the lowest PM2.5 level
fig.add_annotation(
    x=low["hour"],
    y=low["pm25_value"],
    text=f"<b>Lowest: {low['hour']}h</b>",
    showarrow=True,
    arrowhead=2
)

# 5. IMPROVE LAYOUT AND STYLING
# Enhance readability and presentation quality
fig.update_layout(
    # Centered title with styled subtitle
    template="plotly_white",

    # Set font style for entire chart
    font=dict(
        family="Helvetica",
        size=14,
        color="black"
    ),

    # Axis labels
    xaxis_title="Hour of Day",
    yaxis_title="Average PM2.5 (µg/m³)", 

    # Axis formatting
    xaxis=dict(showgrid=False, tickmode='linear'),
    yaxis=dict(showgrid=True, gridcolor="lightgray"),

    # Centered title with styled subtitle
    title={
        'text': "<b>RUSH HOUR POLLUTION: When Delhi's Air Is Most Dangerous</b><br>"
                "<span style='font-size:12px;' >PM2.5 peaks during commuting hours, highlighting the impact of daily urban activity (April 2026)</span>",
        'x': 0.5,
        'xanchor': 'center'
    }
)

# 6. ENHANCE LINE AND MARKER STYLING
# Smooth the line and improve visual emphasis
fig.update_traces(
    line_shape="spline",    # Smooth curves
    line=dict(width=2),     # Thicker line for emphasis
    marker=dict(size=8)     # Larger markers for visibility
)

# 7. ADD VISUAL HIGHLIGHT REGIONS (VRECT)
# Highlight morning peak period (commuting hours)
fig.add_vrect(
    x0=6, x1=9,
    fillcolor="red", opacity=0.1,
    line_width=0,
    annotation_text="<b>Morning Peak</b>",
    annotation_position="top left"
)

# Highlight evening rise period (commuting hours)
fig.add_vrect(
    x0=18, x1=21,
    fillcolor="orange", opacity=0.1,
    line_width=0,
    annotation_text="<b>Evening Rise</b>",
    annotation_position="top left"
)

# 8. DISPLAY THE DATA VISUALIZATION
fig.show()
```
