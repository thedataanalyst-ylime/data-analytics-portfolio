**CORRELATION ANALYSIS WITH NUMERICAL COLUMNS**

```python
# 1. Select Numerical Columns
# Correlation works best on numerical fields.

numeric_cols = hdb.select_dtypes(include='number')

numeric_cols.head()
```

```python
# 2. Compute Correlation Matrix

# This computes Pearson correlation values between:
#-1 → strong negative relationship
# 0 → no relationship
# +1 → strong positive relationship

corr_matrix = numeric_cols.corr()

corr_matrix.head()
```

```python
# 3. Find Strongest Correlations with Resale Price
# Correlation with resale price
price_corr = (
    corr_matrix['resale_price']
    .sort_values(ascending=False)
)

price_corr
```

```python
# 4. View Top Positive Correlations
# “Why does this increase price?”

price_corr.head(40)
```

```python
# 5. View Strongest Negative Correlations
# “Why does this reduce price?”

price_corr.tail(30)
```

```python
# Create a Correlation Heatmap

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(18,12))

sns.heatmap(
    corr_matrix,
    cmap='coolwarm',
    center=0
)

plt.title('Correlation Matrix of HDB Features')
plt.show()
```

```python
# Top 15 strongest correlations with resale_price. Focus on top correlations ONLY.
top_corr = price_corr.abs().sort_values(ascending=False).head(15)

top_corr
```

```python
plt.figure(figsize=(10,6))

top_corr.sort_values().plot(
    kind='barh'
)

plt.title('Top Features Correlated with Resale Price')
plt.xlabel('Correlation Strength')

plt.show()
```

```python
# Find highly correlated pairs
high_corr = (
    corr_matrix
    .abs()
    .unstack()
    .sort_values(ascending=False)
)

high_corr.head(20)
```

**CORRELATION ANALYSIS WITH CATEGORICAL FEATURES**

```python
# WHat is the median resale price per location?

hdb.groupby('planning_area')['resale_price'].median().sort_values()
```

```python
# How does location affect resale price?

planning_area_analysis = (
    hdb.groupby('planning_area')
    .agg({
        'resale_price': ['mean', 'median', 'min', 'max', 'count'],
        'floor_area_sqm': 'mean',
        'hdb_age': 'mean',
        'mrt_nearest_distance': 'mean'
    })
)

planning_area_analysis
```

```python
planning_area_analysis.columns = [
    'avg_price',
    'median_price',
    'min_price',
    'max_price',
    'transaction_count',
    'avg_floor_area',
    'avg_hdb_age',
    'avg_mrt_distance'
]

planning_area_analysis = (
    planning_area_analysis
    .sort_values(by='avg_price', ascending=False)
)

planning_area_analysis
```

**GET A GEOJSON FILE OF SINGAPORE FOR DATA VISUALIZATION**

```python
import requests

# Dataset ID
dataset_id = "d_2cc750190544007400b2cfd5d7f53209"

# Step 1: Get temporary download URL
url = f"https://api-open.data.gov.sg/v1/public/api/datasets/{dataset_id}/poll-download"

response = requests.get(url)
json_data = response.json()

# Error handling
if json_data['code'] != 0:
    print(json_data['errMsg'])
    exit(1)

# Step 2: Retrieve actual download link
download_url = json_data['data']['url']

# Step 3: Download GEOJSON
geojson_response = requests.get(download_url)

# Step 4: Save locally
with open("singapore_planning_area.geojson", "w") as file:
    file.write(geojson_response.text)

print("GeoJSON file downloaded successfully.")
```
