We have the total line item amount (the retail price) as well as the wholesale price of the product.

We want to calculate the **total wholesale line item amount** so we can use it to calculate profit.

Can you combine the existing wholesale amount and quantity columns to create a new column in the `DataFrame` called `total_wholesale_amount` which holds the new values?

```python
#Calculation Logic: current_wholesale_price x quantity

coffee_df['total_wholesale_amount'] = coffee_df['current_wholesale_price'] * coffee_df['quantity']
coffee_df
```

You might have noticed earlier when running the `dtypes` command that the `transaction_date` and `customer_since` columns are currently being treated like strings. We need to convert them into date columns so we can calculate factors like how long someone's been a customer.

Try using pd.to_datetime() to convert these two columns to date representations

```python
#Check datatypes of each column

coffee_df.dtypes
```

```python
#Convert transaction_date and customer_since to datetime

coffee_df['transaction_date'] = pd.to_datetime(coffee_df['transaction_date'])
coffee_df['customer_since'] = pd.to_datetime(coffee_df['customer_since'])

coffee_df
```

##### <font color='green'>**SPEND ANALYSIS for CUSTOMER GENERATION**</font><br>

**COVERS 3 FINANCIAL MEASURES:**<br>
1. **Total Revenue by Generation:** Who brings in the most money overall? <br>
2. **Avg Spend per Transaction:** Who spends more per visit? <br>
3. **Avg Spend per Customer:** Who is most valuable? <br>


```python
# 1. TOTAL REVENUE BY GENERATION
# What this means: Who brings in the most money overall? 
# When to use: For Business scale

revenue_by_generation = (
                            coffee_df
                            .groupby('customer_generation')['line_item_amount']
                            .sum()
                            .sort_values(ascending=False)
                            .reset_index()
)
revenue_by_generation
```

```python
#2. AVERAGE SPEND PER TRANSACTION BY GENERATION
# What this means: Who spends more per visit?
# When to use: For Upselling

avg_txnspend_by_generation = (
                                coffee_df
                                .groupby('customer_generation')['line_item_amount']
                                .mean()
                                .sort_values(ascending=False)
                                .reset_index()
)
avg_txnspend_by_generation
```

```python
#3. AVERAGE SPEND PER CUSTOMER BY GENERATION
# What this means: Who is most valuable?
# When to use: For Loyalty strategy

#Step 1: Compute total spend per customer_generation and customer_id
spend_by_customer = (
                    coffee_df
                    .groupby(['customer_generation', 'customer_id'])['line_item_amount']
                    .sum()
                    .reset_index()
) 
spend_by_customer
```

```python
#Step 2: Compute average spend per customer_generation
avg_spend_by_generation = (
                        spend_by_customer
                        .groupby('customer_generation')['line_item_amount']
                        .mean()
                        .sort_values(ascending=False)
                        .reset_index()
)
avg_spend_by_generation
```

```python
#4. NUMBER OF TRANSACTIONS BY GENERATION
# What this means: Who are frequent customers?
# When to use: For Loyalty strategy

#Step 1: Compute total count of transactions per customer_generation and customer_id
txns_by_customer = (
                    coffee_df
                    .groupby(['customer_generation', 'customer_id'])['transaction_date']
                    .count()
                    .reset_index()
                    .rename(columns={'transaction_date': 'number_of_transactions'})
)
txns_by_customer
```

```python
#Step 2: Compute total count of transactions per customer_generation
txns_by_generation = (
                        txns_by_customer
                        .groupby('customer_generation')['number_of_transactions']
                        .sum()
                        .sort_values(ascending=False)
                        .reset_index()
)
txns_by_generation
```

```python
#2. Create a histogram for profit between -$5 and $15

filtered_df = coffee_df[(coffee_df['profit'] > -5) & (coffee_df['profit'] <= 15)]

#Compute median profit
median_profit = filtered_df['profit'].median()

fig = px.histogram(
    filtered_df,
    x='profit',
    nbins=30,
    opacity=0.9,
    color_discrete_sequence=['#2EBB57'],
    title=(
        "Most Profit Comes from Small, Frequent Transactions"
        "<br><sup>The café’s profitability is driven by consistent everyday purchases, "
        "highlighting the importance of increasing average order value across all customers.</sup>"
    ),
    labels={'profit': 'Profit per Transaction ($)'},
)

# Add median reference line
fig.add_vline(
    x=median_profit,
    line_width=4,
    line_dash="dash",
    line_color='#B22222',

    annotation_text=f"Median Profit = ${median_profit:.2f}",
    annotation_position='top',
    annotation_font_size=14,
    annotation_font_color='#B22222'
)

# Chart style
fig.update_layout(
    template='simple_white',
    bargap=0.1,
    title_x=0.02,
    title_font_size=24,
    xaxis_title='Profit per Transaction ($)',
    yaxis_title='Number of Transactions',
    font=dict(
        family='Helvetica',
        size=14
    ),
    height=600,
    width=1200
)

#Show where most of the transactions are
fig.add_vrect(
    x0=0,
    x1=median_profit,
    fillcolor='lightgreen',
    opacity=0.3,
    line_width=0
)

#Bar Styles
fig.update_traces(
    marker_line_width=1,
    marker_line_color='white'
)

fig.show()
```

```python
#2A. Create a scatterplot to show the relationship between the transaction amount and age

#Step 1: Convert customer_birth_year to date by converting it to integer first.

coffee_df['birth_year'] = pd.to_numeric(
    coffee_df['customer_birth_year'],
    errors='coerce' #invalid parsing will be set as NaN
).astype('Int64')   # keeps NaN safely
```

```python
coffee_df
```

```python
#Check the data type of the new field birth_year >> Int64

coffee_df.info()
```

```python
#Step 2: Convert customer_birth_year to datetime

coffee_df['birth_date'] = pd.to_datetime(
    coffee_df['birth_year'].astype(str) + '-01-01',
    errors='coerce'
)
coffee_df
```

```python
#Step 3: Compute age using a reference date and converts it to integer

reference_date = pd.Timestamp('2019-04-01')

coffee_df['age'] = (
    (reference_date - coffee_df['birth_date'])
    .dt.days // 365
).astype('Int64')

coffee_df
```

```python
#Check the data type of the new field birth_year >> Int64

coffee_df.info()
```

```python
#Step 4: Create a scatterplot to show the relationship between the transaction amount and age for ALL transactions

fig = px.scatter(
    coffee_df,
    x='age',
    y='line_item_amount',
    color='product_group',
    
    labels={
        'age': 'Customer Age',
        'line_item_amount': 'Transaction Amount ($)',
        'product_group': 'Product Group'
    },
    color_discrete_map={
        'Beverages': '#6F4E37',
        'Food': '#D9822B',
        'Whole Bean/Teas': '#2E8B57',
        'Add-Ons': '#C8A97E',
        'Merchandise': '#8B5FBF'
    },
    hover_data={
        'product_group': True, 
        'product_type': True,
        'age': True,
        'line_item_amount': ':.2f'
    },
    
    title=("Age Does Not Drive Spending — Product Choice Does"
           "<br><sup>Most transactions are small and consistent across customer ages, "
           "while rare high-value purchases are driven by specific product types.</sup>"
    )
)

#Chart Style
fig.update_layout(
    template='simple_white',
    font=dict(family='Helvetica', size=14, color='#2B2B2B'),
    title_x=0.02,
    title_font_size=22,
    xaxis_title='Customer Age',
    yaxis_title='Transaction Amount ($)',
    legend_title='Product Group',
    height=600,
    width=1200,
    margin=dict(t=140, l=70, r=40, b=70)
)

#Circle Styles
fig.update_traces(
    marker=dict(
        size=10,
        opacity=0.8,
        line=dict(width=0.2, color='white')
    )
)

#Add annotation to show higher-value purchases 
fig.add_annotation(
    x=52,
    y=320,
    text='Higher-value purchases are driven<br>more by product type than age',
    showarrow=False,
    font=dict(family='Helvetica', size=15, color='#2B2B2B')
)
fig.show()
```
